```python
import os
import urllib.request
from tqdm import tqdm

# 1. Pfade und Dropbox-Link definieren
target_folder = "/kaggle/working/program_ml/assets/datasets/DRFL/JW3"
# Ersetzen Sie diesen Link mit Ihrem echten Dropbox-Freigabelink
dropbox_url = "https://www.dropbox.com/scl/fi/ba39xuv8zdvcv7vc360a8/violets_dataset_1782855348228.wav?rlkey=wnapomikweylcfxuxx02trljk&st=zovu4588&dl=0" 

# 2. Ordner erstellen, falls er noch nicht existiert
os.makedirs(target_folder, exist_ok=True)

# 3. Dropbox-Link für den Direkt-Download anpassen (dl=0 zu dl=1 ändern)
if "dl=0" in dropbox_url:
    download_url = dropbox_url.replace("dl=0", "dl=1")
elif "dl=1" not in dropbox_url:
    download_url = dropbox_url + "&dl=1" if "?" in dropbox_url else dropbox_url + "?dl=1"
else:
    download_url = dropbox_url

# Dateinamen aus der URL extrahieren (Bereinigt URL-Parameter)
filename = download_url.split('/')[-1].split('?')[0]
output_path = os.path.join(target_folder, filename)

# 4. Download-Klasse mit Fortschrittsanzeige (wichtig für große .wav-Dateien)
class DownloadProgressBar(tqdm):
    def update_to(self, b=1, bsize=1, tsize=None):
        if tsize is not None:
            self.total = tsize
        self.update(b * bsize - self.n)

print(f"Downloading: {filename}")
print(f"Output: {output_path}")

try:
    with DownloadProgressBar(unit='B', unit_scale=True, miniters=1, desc="Downloading") as t:
        urllib.request.urlretrieve(download_url, filename=output_path, reporthook=t.update_to)
    print("\n✅ Download Finished ")
except Exception as e:
    print(f"\n❌ Error Please Check the Link: {e}")
```
